```python
# deep learning--it is branch of ML ,it uses ANN to learn the patterns from large data
# How Deep Learning Works
# A Deep Neural Network (DNN) consists of multiple layers.
# 1. Input Layer -- Recieves raw data--activation --relu
# 2. Hidden Layers-- Perform calculations and extract features
# 3. Outupt Layer -- Produces Final Prediction

# Training Process
# Deep Learning models learn using three main steps:

# 1 Forward Propagation
# # Data passes thorough the network and produces a prediction
# 2️ Loss Calculation
# The model checks how wrong the prediction is.
# 3️ Backpropagation
# The model adjusts weights using algorithms like Gradient Descent to reduce errors.
# This process repeats thousands or millions of times until the model learns patterns

#NLP---Natural Language Processing
# Natural Language Processing (NLP) is a field of Artificial Intelligence that enables computers to understand, 
# interpret, and generate human language.
#nltk---natural language toolkit---library used for nlp in python

#6 steps--preprocessing steps using nltk
#1--lower()
#2---re()---remove all the special characters, punctuations
#3---tokenization---converting sentence to words
#4---stop word removal---(ex--is, the, that, are, were)
#5---stemming or lemmatization
#----stemming---(convert to root word--ex--playing -> play,it may create words which is not in dictionary)
#-----lemmatization----(convert to root word--ex--playing -> play, uses proper dictionary format)
#6----vectorization-----converting to numeric values

# "This Product is AMAZING!!! Totally worth it "

# lowercase → this product is amazing!!! totally worth it
# remove special → this product is amazing totally worth it
# tokenize → ["this","product","is","amazing","totally","worth","it"]
# remove stopwords → ["product","amazing","worth"]
# lemmatize → ["product","amazing","worth"]
# vectorize → numerical vector
```

```python
import pandas as pd
import numpy as np
import re
import nltk
import string

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
```

```python
df.drop('category',inplace=True,axis=1)
df.drop('rating',inplace=True,axis=1)
```

```python
#converting to lowercase
df['text_']=df['text_'].str.lower()
df
```

```python
# removing special characters, punctuations etc
import re

df['text_'] = df['text_'].apply(lambda x: re.sub(r'[^a-zA-Z\s]', '', x))
df
                                            
```

```python
nltk.download('punkt')
```

```python
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize
df['token']=df['text_'].apply(word_tokenize)
df
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
# removing stopwords
stop_words=set(stopwords.words('english'))
df['token']=df['token'].apply(lambda x:[word for word in x 
                              if word not in stop_words])
print(df)
```

```python
# lemmatization
lemmatizer=WordNetLemmatizer()
df['token']=df['token'].apply(lambda x:[lemmatizer.lemmatize(word) for word in x])
print(df)
```

```python
# join back tokens to string
df['clean_text']=df['token'].apply(lambda x:' '.join(x))
df
```

```python
#vectorization
tfid=TfidfVectorizer()
x=tfid.fit_transform(df['clean_text'])
```

```python
df['label']=df['label'].map({'OR':0,'CG':1})
y=df['label'].values
```

```python

y.shape
```

```python
# splitting the data for training and testing
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print(f"Training samples: {X_train.shape[0]}")
print(f"Testing samples: {X_test.shape[0]}")
```

```python

# Logistic Regression

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

lr_model = LogisticRegression()
lr_model.fit(X_train, y_train)

lr_pred = lr_model.predict(X_test)

print("=== Logistic Regression ===")
print(f"Accuracy: {accuracy_score(y_test, lr_pred):.4f}")
print(classification_report(y_test, lr_pred, target_names=['OR (Real)', 'CG (Fake)']))
```

```python

# Naive Bayes

from sklearn.naive_bayes import MultinomialNB

nb_model = MultinomialNB()
nb_model.fit(X_train, y_train)

nb_pred = nb_model.predict(X_test)

print("=== Naive Bayes ===")
print(f"Accuracy: {accuracy_score(y_test, nb_pred):.4f}")
print(classification_report(y_test, nb_pred, target_names=['OR (Real)', 'CG (Fake)']))
```

```python

# Random Forest

from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)

print("=== Random Forest ===")
print(f"Accuracy: {accuracy_score(y_test, rf_pred):.4f}")
print(classification_report(y_test, rf_pred, target_names=['OR (Real)', 'CG (Fake)']))
```

```python
# ============================================================
# MODEL 4: Deep Learning (ANN) - Memory Efficient Version
# ============================================================
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
import scipy.sparse as sp

# ✅ DO NOT convert to dense — use a sparse-compatible generator instead
def sparse_batch_generator(X, y, batch_size=64):
    num_samples = X.shape[0]
    indices = np.arange(num_samples)
    while True:
        np.random.shuffle(indices)
        for start in range(0, num_samples, batch_size):
            batch_idx = indices[start:start + batch_size]
            X_batch = X[batch_idx].toarray()   # only one batch at a time in memory
            y_batch = y[batch_idx]
            yield X_batch, y_batch

# Split sparse matrix directly (no .toarray())
X_train_sp, X_test_sp, y_train_sp, y_test_sp = train_test_split(
    x, y, test_size=0.2, random_state=42
)

input_dim = X_train_sp.shape[1]
batch_size = 64
steps_per_epoch = X_train_sp.shape[0] // batch_size

ann_model = Sequential([
    Dense(128, activation='relu', input_shape=(input_dim,)),
    Dropout(0.3),
    Dense(64, activation='relu'),
    Dropout(0.3),
    Dense(1, activation='sigmoid')
])

ann_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
ann_model.summary()

early_stop = EarlyStopping(monitor='loss', patience=3, restore_best_weights=True)

# Train using generator — only one batch loaded into RAM at a time
history = ann_model.fit(
    sparse_batch_generator(X_train_sp, y_train_sp, batch_size=batch_size),
    steps_per_epoch=steps_per_epoch,
    epochs=20,
    callbacks=[early_stop],
    verbose=1
)
```

```python
# Evaluate — again batch by batch to avoid memory error
def predict_in_batches(model, X_sparse, batch_size=512):
    preds = []
    for start in range(0, X_sparse.shape[0], batch_size):
        batch = X_sparse[start:start + batch_size].toarray()
        preds.extend((model.predict(batch, verbose=0) > 0.5).astype(int).flatten())
    return np.array(preds)

ann_pred = predict_in_batches(ann_model, X_test_sp)

# Accuracy
ann_acc = accuracy_score(y_test_sp, ann_pred)
print("=== Deep Learning (ANN) ===")
print(f"Accuracy: {ann_acc:.4f}")
print(classification_report(y_test_sp, ann_pred, target_names=['OR (Real)', 'CG (Fake)']))
```

```python
# Update comparison table with correct test labels
results = {
    'Model': ['Logistic Regression', 'Naive Bayes', 'Random Forest', 'ANN (Deep Learning)'],
    'Accuracy': [
        accuracy_score(y_test, lr_pred),
        accuracy_score(y_test, nb_pred),
        accuracy_score(y_test, rf_pred),
        ann_acc
    ]
}

results_df = pd.DataFrame(results)
results_df['Accuracy (%)'] = (results_df['Accuracy'] * 100).round(2)
print(results_df)

plt.figure(figsize=(9, 5))
plt.bar(results_df['Model'], results_df['Accuracy (%)'], color=['steelblue','orange','green','crimson'])
plt.ylim(80, 100)
plt.title('Model Accuracy Comparison - Fake Review Detection')
plt.ylabel('Accuracy (%)')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python

# PREDICT FUNCTION — Using ANN Model


def predict_review(review_text):
    # Step 1: Lowercase
    review = review_text.lower()
    
    # Step 2: Remove special characters
    review = re.sub(r'[^a-zA-Z\s]', '', review)
    
    # Step 3: Tokenize
    tokens = word_tokenize(review)
    
    # Step 4: Remove stopwords
    tokens = [w for w in tokens if w not in stop_words]
    
    # Step 5: Lemmatize
    tokens = [lemmatizer.lemmatize(w) for w in tokens]
    
    # Step 6: Join back to string
    clean = ' '.join(tokens)
    
    # Step 7: Vectorize using the SAME fitted TF-IDF
    vec = tfid.transform([clean]).toarray()  # single row, safe to convert
    
    # Step 8: Predict
    prob = ann_model.predict(vec, verbose=0)[0][0]
    label = 1 if prob > 0.5 else 0
    confidence = prob if label == 1 else 1 - prob

    # Step 9: Display Result
    print("=" * 55)
    print(f"  Review   : {review_text[:80]}{'...' if len(review_text) > 80 else ''}")
    print("-" * 55)
    if label == 1:
        print(f"  Result   : ❌ CG — Computer Generated (Fake Review)")
    else:
        print(f"  Result   : ✅ OR — Original (Genuine Review)")
  
    print("=" * 55)
    
    return label, confidence




```

```python

predict_review("SHEET COLOR IS NICE BUY FOR MY SIZE FOR MY TOWN")
```
